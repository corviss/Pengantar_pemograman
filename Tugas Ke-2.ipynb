```python
import pandas as pd
import matplotlib.pyplot as plt

my_dataset = pd.read_excel(
    'Smith_glass_post_NYT_data.xlsx',
    sheet_name='Supp_majors',
    engine='openpyxl')

sub_data1 = my_dataset[my_dataset.Na2O > 4]
sub_data2 = my_dataset[my_dataset.Na2O < 4]

fig, ax = plt.subplots()

Dots = ["two","three-b"]
colors = ["blue","red","green","orange"]
x_line = (3.1, 4)
y_line = (62, 60.2)

i = 0

for dot in Dots:

    my_dot1 = sub_data1[sub_data1.Epoch == dot]
    ax.scatter(
        my_dot1.Na2O,
        my_dot1.SiO2,
        marker="o",
        facecolors="none",
        edgecolors=colors[i],
        linewidths=1.2,
        label=f"$Na_2O$ > 4, Epoch = {dot}")
    i += 1

    my_dot2 = sub_data2[sub_data2.Epoch == dot]
    ax.scatter(
        my_dot2.Na2O,
        my_dot2.SiO2,
        marker="o",
        facecolors="none",
        edgecolors=colors[i],
        linewidths=1.2,
        label=f"$Na_2O$ < 4, Epoch = {dot}")
    i += 1

ax.set_title("Random S")
ax.set_xlabel("$Na_2O$ Something-something")
ax.set_ylabel("$SiO_2$ Something-something")
ax.plot(x_line, y_line, color = "0")
ax.annotate(r"Random Line", xy = (3.6, 61), xytext=(3.4, 62.5), arrowprops=dict(arrowstyle="->", connectionstyle="arc3"))
ax.text(3.2, 59.5, r"Upside down", rotation=180)
ax.minorticks_on()
ax.legend(framealpha=0.1)
```
