```python
import pandas as pd
import matplotlib.pyplot as plt

my_dataset1 = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces')

x = my_dataset1.Zr
y = my_dataset1.Th

fig, ax = plt.subplots() 

my_data1 = my_dataset1[(my_dataset1.Epoch == 'one')]  #melakukan filter/pemisahan yang memisahkan data dengan kolom epoch yang bernilai "one"
ax.scatter(my_data1.Zr, my_data1.Th, label='Epoch 1') # "my_data.Zr" merupakan penentuan sumbu/axis x menggunakan variabel "my_data" khusus pada kolom Zr
                                                      # "my_data.Th" melakukan proses yang sama pada "my_data.Zr" namun munggunakan Th pada sumbu y
                                                      # "label" memberikan dengan value "Epoch 1"

my_data2 = my_dataset1[(my_dataset1.Epoch == 'two')]
ax.scatter(my_data2.Zr, my_data2.Th, label='Epoch 2')

my_data3 = my_dataset1[(my_dataset1.Epoch == 'three')]
ax.scatter(my_data3.Zr, my_data3.Th, label='Epoch 3')

my_data4 = my_dataset1[(my_dataset1.Epoch == 'three-b')]
ax.scatter(my_data4.Zr, my_data4.Th, label='Epoch 3b')
    
ax.set_title("My Third Diagram")   
ax.set_xlabel("Zr [ppm]")
ax.set_ylabel("Th [ppm]")
ax.legend()
```

```python
# for loop digunakan untuk mengulang sebuah urutan (string, tuple, list, set)
# list []  = memungkinkan untuk mempumyuai lebih dari satu. muteable (bisa berubah) dan flexsible
# tuple () = immutable tapi lebih cepat diakses
epochs = ['one','two','three','three-b'] # pembuatan variable dengan list []

fig, ax = plt.subplots() # buat gambar grafik/diagram
for epoch in epochs:     # "for" digunakan agar variable "epochs" diproses satu persatu
    my_data = my_dataset1[(my_dataset1.Epoch == epoch)]         # melakukan filter(pemisahan) data pada file excel berdasarkan epoch
    ax.scatter(my_data.Zr, my_data.Th, label="Epoch " + epoch)  # "my_data.Zr" merupakan penentuan sumbu/axis x menggunakan variabel "my_data" khusus pada kolom Zr
                                                                # "my_data.Th" melakukan proses yang sama pada "my_data.Zr" namun munggunakan Th pada sumbu y
                                                                # "label" memberikan label dengan tulisan (Epoch) + value dari variable "epoch"
ax.set_title("My Third Diagram again")
ax.set_xlabel("Zr [ppm]")
ax.set_ylabel("Th [ppm]")
ax.legend()
```

```python
def ec(f, d, c0):           # mendefinisikan sebuah persamaan dengan nama "ec"
    cl = c0/(d * (1-f) + f) # isi/perhitungan yang ada pada persamaan atau persamaan "ec"
    return cl               # membuat hasil dari persamaan(cl) menjadi namanya yang awal("ec")

my_c = ec(f=0.5, d=0.1, c0=100) # memberikan nilai dari "(f, d, c0)" 
print(type(my_c))

print('RESULT: '+ str(int(my_c)) + ' ppm') # "str(int(my_c))" mengubah tipe data "my_c" dari float ke integer kemudian ke string
```

```python
import matplotlib.pyplot as plt
import numpy as np

def ec(f, d, c0):
    cl = c0/(d * (1-f) + f)
    return cl

my_f = np.linspace(0.3,1, 8)

# proses membedakan data dengan nilai "d" yang berbeda
my_c1 = ec(f=my_f, d =0.1, c0=100) 
my_c2 = ec(f=my_f, d=1, c0=100)
my_c3 = ec(f=my_f, d=2, c0=100)

# Proses membuat gambar grafik berdasarkan data yang dibedakan diatas
fig, ax = plt.subplots() 
ax.plot(my_f, my_c1, label="Eq cryst. D = 0.1")
ax.plot(my_f, my_c2, label="Eq cryst. D = 1")
ax.plot(my_f, my_c3, label="Eq cryst. D = 2")

ax.set_xlabel('F')
ax.set_ylabel('C [ppm]')
ax.legend()
```

```python
import matplotlib.pyplot as plt
import numpy as np

def ec(f, d, c0):
    cl = c0/(d * (1-f) + f)
    return cl

my_f = np.linspace(0.3,1, 8)

d = [0.1, 1, 2]

fig, ax = plt.subplots() 

# menggunakan for loop untuk mengurangi penggunaan code seperti pada sebelumnya
for my_d in d:
    my_c = ec(f=my_f, d=my_d, c0=100)
    ax.plot(my_f, my_c, label='Eq cryst. D = ' +  str(my_d))


ax.set_xlabel('F')
ax.set_ylabel('C [ppm]')
ax.legend()
```

```python
# mengimport submodule yang benar dari libary matplotlib

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import cm            # cm stands for colormap

my_data = pd.read_csv('DEM.csv')
fig, ax = plt.subplots() 
# membuat plot sebar 
# ".values" memastikan agar data pada kolum diubah ke bentuk array numpy
ax.scatter(x = my_data.X_LOC.values,     # sumbu x dengan data dari kolom "X_LOC"
           y = my_data.Y_LOC.values,     # sumbu y dengan data dari kolom "Y_LOC"
           c = my_data.ELEVATION.values, # memberikan warna pada titik kordinat berdasarkan kolom
           s = 2,                        # menentukan ukuran titik kordinat (2 = angka yang sangat  kecil)
           cmap = 'hot',                 # mengubah plot sebar(scatter plot) menjadi representasi gradient(map dengan gradient warna)
           linewidth = 0,                # membuat garis pinggir pada titik menjadi tidak ada
           marker = 'o')                 # mengubah bentuk titik
ax.axis('equal')                         # membuat panjang  sumbu x dan sumbu y sama panjang
ax.axis('off')                           # membuat sumbu dan penanda-nya menghilang

# membuat colorbar
colorbar = fig.colorbar(cm.ScalarMappable(cmap='hot'), # membuat colorbar dengan warna dari "hot" colormap
                        extend='max',                  # membuat tanda panah apabila ada data yang melewati range dari colormap
                        ax=ax)                         # menentukan penempatan colorbar
# penesuaian colorbar
colorbar.set_label('Elevation [m]',                    # menentukan judul/label colormap
                    rotation=270,                      # menentukan arah dari label
                    labelpad=20)                       # menentukan jarak label dari colorbar
colorbar.outline.set_edgecolor('Grey')                 # menentukan warna garis pinggir dari colorbar
```
