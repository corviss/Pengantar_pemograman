```python
# Menginstall/memasang module yang cocok dan diperlukan
import pandas as pd
from scipy.stats.mstats import gmean, hmean
import matplotlib.pyplot as plt

# Membaca dan menentukan file data yang diperlukan
my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces')

a_mean = my_dataset.Zr.mean()     # Untuk membuat aritmathic mean dari data
g_mean = gmean(my_dataset['Zr'])  # Untuk geometric mean dari data
h_mean = hmean(my_dataset['Zr'])  # Untuk membuat harmonic mena dari data

# Membuat display agar data dapat disajikan
print ('-------')
print ('arithmetic mean')
print ("{0:.1f} [ppm]".format(a_mean))
print ('-------')

print ('geometric mean')
print ("{0:.1f} [ppm]".format(g_mean))
print ('-------')

print ('harmonic mean')
print ("{0:.1f} [ppm]".format(h_mean))
print ('-------')

# Membuat grafik
fig, ax = plt.subplots()
ax.hist(my_dataset.Zr,              # Untuk menentukan data apa yang digunakan pada pembuatan histogram
        bins='auto',                # Untuk menentukan bins(pengelompokan data pada satu balok/imterval) pada grafik
        density=True,               # Untuk membuat sifat sumbu y menjadi probabilitas(bukan sifat menunjukkan jumlah)
        edgecolor='k',              # Untuk menentukan warna garis pinggir balok
        label='Measurements Hist',  # Untuk memberikan label
        alpha=0.8)                  # Untuk menentukan sifat transparansi warna pada balok
# "ax.axvline" untuk membuat garis pada sumbu x 
ax.axvline(a_mean,                  #ax1 Untuk menentukan data apa yang digunakan pada pembuatan histogram
           color='purple',          #ax2 Untuk menentukan warna garis
           label='Arithmetic mean', #ax3 Untuk memberikan label
           linewidth=2)             #ax4 Untuk menetukan lebar/ukuran garis                  
ax.axvline(g_mean,                  #ax1
           color='orange',          #ax2
           label='Geometric mean',  #ax3
           linewidth=2)             #ax4
ax.axvline(h_mean,                  #ax1
           color='green',           #ax2
           label='Harmonic mean',   #ax3
           linewidth=2)             #ax4
ax.set_xlabel('Zr [ppm]')
ax.set_ylabel('Probability density')
ax.legend()
```

```python
# Menginstall/memasang module yang cocok dan diperlukan
import pandas as pd
import matplotlib.pyplot as plt

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces')

variance =  my_dataset['Zr'].var()  # Untuk menentukan variansi/penyebaran data
stddev =  my_dataset['Zr'].std()    # Untuk menentukan standar deviasi data

# Membuat display agar data dapat disajikan
print ('-------')
print ('Variance')
print("{0:.0f} [square ppm]".format(variance))
print ('-------')
print ('Standard Deviation')
print("{0:.0f} [ppm]".format(stddev))
print ('-------')

fig, ax = plt.subplots()
ax.hist(my_dataset.Zr, bins= 20, density = True, edgecolor='k', label='Measurements Hist') 
ax.axvline(my_dataset['Zr'].mean()-stddev, color='purple', label=r'mean - 1$\sigma$', linewidth=2)                                    # menentukan batas bawah dari rata-rata
ax.axvline(my_dataset['Zr'].mean()+stddev, color='green', label=r'mean + 1$\sigma$', linewidth=2)                                     # menentukan batas atas dari rata-rata
ax.axvspan(my_dataset['Zr'].mean()-stddev, my_dataset['Zr'].mean()+stddev, alpha=0.5, color='orange', label=r'mean $\pm$ 1$\sigma$')  # menunjukkan area dengan kurang dan lebih dari 1 tsandar deviasi
ax.set_xlabel('Zr [ppm]')
ax.set_ylabel('Probability density')
ax.legend()
```

```python
# Menginstall/memasang module yang cocok dan diperlukan
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

my_dataset = pd.read_excel('Smith_glass_post_NYT_data.xlsx', sheet_name='Supp_traces')

# np.percentile digunakan untuk menghintung persentile dari array data
q1 = np.percentile(my_dataset.Zr, 25,   # untuk menentukan quartil pertama(25) 
                   method = 'midpoint') # untuk menghitung rata-rata titik quaartil
q3 = np.percentile(my_dataset.Zr, 75, 
                   method = 'midpoint')
  
iqr = q3 - q1 # Interquaritle range (IQR)  

print ('-------')
print ('Interquaritle range (IQR): {0:.0f} [ppm]'.format(iqr))
print ('-------')

fig, ax = plt.subplots()
ax.hist(my_dataset.Zr, bins='auto', density=True, edgecolor='k', label='Measurements Hist') 
ax.axvline(q1, color='purple', label='Q1', linewidth=2)
ax.axvline(q3, color='green', label='Q3', linewidth=2)
ax.axvspan(q1, q3, alpha=0.1, color='orange', label='Interquaritle range (IQR)')
ax.set_xlabel('Zr [ppm]')
ax.set_ylabel('Probability density')
ax.legend()
```
