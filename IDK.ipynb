```python
import numpy as np

# Mendefinisikan fungsi "integrate_rec"
def integrate_rec(f, a, b, n):
    # Implementation of the rectanlge method 
    h = (b-a)/n
    x = np.linspace(a, b, n+1) 
    i=0
    area=0
    while i<n:
        sup_rect =  f(x[i])*h
        area += sup_rect
        i += 1 
    return area

# "np.sin"(sin) dan "np.pi"(pi) merupakan sebuah nilai dalam library numpy
sup_5 = integrate_rec(np.sin, 0, np.pi/2, 5)
sup_10 = integrate_rec(np.sin, 0, np.pi/2, 10)
sup_100 = integrate_rec(np.sin, 0, np.pi/2, 100)

print('Using n=5, the rectangle method returns a value of {:.2f}'.format(sup_5))
print('Using n=10, the rectangle method returns a value of {:.2f}'.format(sup_10))
print('Using n=100, the rectangle method returns a value of {:.2f}'.format(sup_100))
```

```python
import numpy as np

# Mendefinisikan fungsi "integrate_trap"
def integrate_trap(f, a, b, n):
    # Implementation of the trapezoidal rule
    h = (b-a)/n
    x = np.linspace(a, b, n+1) 
    i=1
    area = h*(f(x[0]) + f(x[n]))/2
    while i<n:
        sup_rect = f(x[i])*h
        area += sup_rect 
        i += 1 
    return area

# "np.sin"(sin) dan "np.pi"(pi) merupakan sebuah nilai dalam library numpy
sup_5 = integrate_trap(np.sin, 0, np.pi/2, 5)
sup_10 = integrate_trap(np.sin, 0, np.pi/2, 10)

# menyajikan/memberikan data
print('Using n=5, the trapezoidal rule returns a value of {:.2f}'.format(sup_5))
print('Using n=10, the trapezoidal rule returns a value of {:.2f}'.format(sup_10))
```

```python
import numpy as np
from scipy import integrate


x = np.linspace(0,9, 3) # 3 divisions %*\textcolor{codegreen}{$[x_0,x_1,x_2]$}*), n=2 
y = x**2

sup_trapz = integrate.trapz(y,x)  # mengolah data dengan metode trapesium
sup_simps = integrate.simps(y,x)  # mengolah data dengan metode simpson


print('Using n=2, the trapezoidal rule returns a value of {:.0f}'.format(sup_trapz))
print('Using n=2, the composite Simpson rule returns a value of {:.0f}'.format(sup_simps))
```

```python
import numpy as np
from scipy.integrate import trapz
import matplotlib.pyplot as plt

r = np.linspace(1,6400,6400)

# medefinisikan fungsi
def density(): # "np.linspace" membuat data dengan interval merata (start, stop, num/jumlah)
    ro_inner_core = np.linspace(13100, 12800, 1220)
    ro_outer_core = np.linspace(12200, 9900, 2259)
    ro_lower_mantle = np.linspace(5600,4400,2171)
    ro_upper_mantle = np.linspace(4130,3400,720)
    ro_crust = np.linspace(3100,2700,30)
   
   # "np.concatenate" digunakan untuk menggabungkan beberapa array menjadi satu
    ro_final = np.concatenate((ro_inner_core, ro_outer_core, ro_lower_mantle, ro_upper_mantle, ro_crust))
    
    return ro_final

ro = density()

#pembuatan grafik
fig, ax = plt.subplots()
ax.plot(r,ro, label="Densities on the Earth's Interior")
ax.set_ylabel(r"Density [Kg/m$^3$]")
ax.set_xlabel("Distance from the Earth center r [km]")
ax.legend()
ax.grid()
```

```python

def gravity(r):
    
    g = np.zeros(len(r))     # "np.zeroes" mengubah nilai array menjadi 0 "len(r)" menghitung jumlah data dalam array. 
    Gr = 6.67408e-11
    r = r * 1000             # from Km to m
    
    for i in range(1,len(r)):
        
        r1 = r[0:i]
        ro1  = ro[0:i]
        r2 = r1[i-1]
    
        y = ro1*r1**2
        y_int = trapz(y,r1)
        
        g1 = ((4 * np.pi*Gr)/(r2**2)) * y_int
        g[i] = g1
           
    return g

g = gravity(r)

fig, ax = plt.subplots()
ax.plot(r,g)
ax.grid()
ax.set_ylabel(r'g [m/s$^2]$')
ax.set_xlabel('Distance from the Earth center r [km]')
```

```python
def pressure(r, ro, g):
    
    p = np.zeros(len(r))
    r = r *1000
    
    for i in range(0,len(r)):    
        r1 = r[i:len(r)]
        ro1 = ro[i:len(r)]
        g1 = g[i:len(r)]
        y = ro1*g1
        p1 = trapz(y,r1)
        p[i] = p1
    return p
    
p = pressure(r,ro,g)/1e9 # expressed in GPa
z = np.linspace(6400, 1, 6400)

fig, ax = plt.subplots()  
ax.plot(z,p)
ax.grid()
ax.set_ylabel('P [GPa]')
ax.set_xlabel('Depth z from the Earth Surface [km]')
```
